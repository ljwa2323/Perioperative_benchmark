```python
from pathlib import Path
import pandas as pd

data_dir = Path("data")
lab_stats = pd.read_csv(data_dir / "lab_stats.csv").rename(
    columns=lambda c: c if c == "op_id" else f"lab_{c}"
)
vit_stats = pd.read_csv(data_dir / "vit_stats.csv").rename(
    columns=lambda c: c if c == "op_id" else f"vit_{c}"
)
ward_vit_stats = pd.read_csv(data_dir / "ward_vit_stats.csv").rename(
    columns=lambda c: c if c == "op_id" else f"ward_vit_{c}"
)
operations = pd.read_csv(data_dir / "operation_.csv")
```

```python
for statistics in (lab_stats, vit_stats, ward_vit_stats):
    operations = operations.merge(statistics, on="op_id", how="left", validate="one_to_one")

stat_columns = [
    column
    for statistics in (lab_stats, vit_stats, ward_vit_stats)
    for column in statistics.columns
    if column != "op_id"
]
operations[stat_columns] = operations[stat_columns].fillna(0)
operations.shape
```

```python
output_path = data_dir / "operation_imputed.csv"
operations.to_csv(output_path, index=False)
output_path
```
