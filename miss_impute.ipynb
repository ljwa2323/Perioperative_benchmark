```r

rm(list = ls()); gc()

# Load necessary libraries for data manipulation and analysis
packages <- c("readxl", "data.table", "dplyr", "magrittr", "lubridate", "stringr", "mice", "parallel", "pbapply", "jsonlite")

for (pkg in packages) {
  if (!require(pkg, character.only = TRUE)) {
    install.packages(pkg, dependencies = TRUE)
    library(pkg, character.only = TRUE)
  }
}
```

```r

lab_stats <- fread("/home/luojiawei/inspire_benchmark_data/lab_stats.csv")
colnames(lab_stats) <- ifelse(colnames(lab_stats) == "op_id", "op_id", paste0("lab_", colnames(lab_stats)))
vit_stats <- fread("/home/luojiawei/inspire_benchmark_data/vit_stats.csv")
colnames(vit_stats) <- ifelse(colnames(vit_stats) == "op_id", "op_id", paste0("vit_", colnames(vit_stats)))
ward_vit_stats <- fread("/home/luojiawei/inspire_benchmark_data/ward_vit_stats.csv")
colnames(ward_vit_stats) <- ifelse(colnames(ward_vit_stats) == "op_id", "op_id", paste0("ward_vit_", colnames(ward_vit_stats)))
operation_ <- fread("/home/luojiawei/inspire_benchmark_data/operation_.csv")

```

```r
operation_ <- merge(operation_, lab_stats, by = "op_id", all.x=T)
operation_ <- merge(operation_, vit_stats, by = "op_id", all.x=T)
operation_ <- merge(operation_, ward_vit_stats, by = "op_id", all.x=T)
```

```r
dim(operation_)
```

```r
lab_item <- names(lab_stats)[names(lab_stats) != "op_id"]
vit_item <- names(vit_stats)[names(vit_stats) != "op_id"]
ward_vit_item <- names(ward_vit_stats)[names(ward_vit_stats) != "op_id"]
```

```r
vit_item
```

```r
# 获取 ward_vit_item、lab_item 和 vit_item 中的每一个列名
ward_vit_cols <- names(operation_)[names(operation_) %in% ward_vit_item]
lab_cols <- names(operation_)[names(operation_) %in% lab_item]
vit_cols <- names(operation_)[names(operation_) %in% vit_item]

# 对这些列进行处理，将 NA 替换为 0
for (col in c(ward_vit_cols, lab_cols, vit_cols)) {
  operation_[is.na(get(col)), (col) := 0]
}
```
